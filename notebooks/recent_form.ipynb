```python
import pandas as pd
import numpy as np

features_path = r"C:\stuff\DATA MINING PROJECT\data\processed\ipl\player_season_features.csv"

df = pd.read_csv(features_path)

print("Shape:", df.shape)
print("Players:", df["player"].nunique())
print("Seasons:", sorted(df["season"].unique()))

display(df.head())
```

```python
# ============================================
# LATEST SEASON PERFORMANCE
# ============================================

latest_season = df["season"].max()

latest = (
    df[df["season"] == latest_season]
    .copy()
)

print("Latest season:", latest_season)
print("Players in latest season:", latest["player"].nunique())

display(latest.head(20))
```

```python
# ============================================
# LATEST SEASON FEATURES
# ============================================

latest_features = latest[
    [
        "player",
        "runs",
        "strike_rate",
        "fours",
        "sixes",
        "boundary_percentage",
        "wickets",
        "economy",
        "bowling_strike_rate"
    ]
].copy()

latest_features = latest_features.rename(
    columns={
        "runs": "latest_runs",
        "strike_rate": "latest_strike_rate",
        "fours": "latest_fours",
        "sixes": "latest_sixes",
        "boundary_percentage": "latest_boundary_percentage",
        "wickets": "latest_wickets",
        "economy": "latest_economy",
        "bowling_strike_rate": "latest_bowling_strike_rate"
    }
)

display(latest_features.head(20))
```

```python
# ============================================
# 3-SEASON RECENT PERFORMANCE
# ============================================

recent_3 = (
    df
    .sort_values(["player", "season"])
    .groupby("player")
    .tail(3)
)

print("Rows:", len(recent_3))
print("Players:", recent_3["player"].nunique())

display(
    recent_3
    .sort_values(["player", "season"])
    .head(20)
)
```

```python
# ============================================
# 3-SEASON AGGREGATE FEATURES
# ============================================

recent_3_features = (
    recent_3
    .groupby("player", as_index=False)
    .agg(
        recent_3_seasons=("season", "nunique"),

        recent_3_runs=("runs", "sum"),
        recent_3_balls=("batting_balls", "sum"),
        recent_3_fours=("fours", "sum"),
        recent_3_sixes=("sixes", "sum"),

        recent_3_wickets=("wickets", "sum"),
        recent_3_bowling_balls=("bowling_balls", "sum"),
        recent_3_runs_conceded=("runs_conceded", "sum")
    )
)

# 3-season batting strike rate
recent_3_features["recent_3_strike_rate"] = (
    recent_3_features["recent_3_runs"] /
    recent_3_features["recent_3_balls"].replace(0, np.nan)
) * 100

# 3-season bowling economy
recent_3_features["recent_3_economy"] = (
    recent_3_features["recent_3_runs_conceded"] /
    (recent_3_features["recent_3_bowling_balls"] / 6).replace(0, np.nan)
)

# 3-season bowling strike rate
recent_3_features["recent_3_bowling_strike_rate"] = (
    recent_3_features["recent_3_bowling_balls"] /
    recent_3_features["recent_3_wickets"].replace(0, np.nan)
)

display(recent_3_features.head(20))
```

```python
print("Shape:", recent_3_features.shape)

print("\nMissing values:")
display(recent_3_features.isna().sum())
```

```python
# ============================================
# PERFORMANCE TREND
# ============================================

df_sorted = df.sort_values(["player", "season"]).copy()

first_last = (
    df_sorted
    .groupby("player")
    .agg(
        first_season=("season", "min"),
        latest_season=("season", "max"),
        first_runs=("runs", "first"),
        latest_runs_from_history=("runs", "last"),
        first_strike_rate=("strike_rate", "first"),
        latest_strike_rate_from_history=("strike_rate", "last"),
        first_wickets=("wickets", "first"),
        latest_wickets_from_history=("wickets", "last"),
        first_economy=("economy", "first"),
        latest_economy_from_history=("economy", "last")
    )
    .reset_index()
)

# Batting run trend
first_last["runs_trend"] = (
    first_last["latest_runs_from_history"] -
    first_last["first_runs"]
)

# Strike-rate trend
first_last["strike_rate_trend"] = (
    first_last["latest_strike_rate_from_history"] -
    first_last["first_strike_rate"]
)

# Wicket trend
first_last["wickets_trend"] = (
    first_last["latest_wickets_from_history"] -
    first_last["first_wickets"]
)

# Economy trend
# Lower economy is better, so improvement is represented as positive
first_last["economy_trend"] = (
    first_last["first_economy"] -
    first_last["latest_economy_from_history"]
)

display(first_last.head(20))
```

```python
# ============================================
# BATTING CONSISTENCY
# ============================================

batting_consistency = (
    df
    .groupby("player", as_index=False)
    .agg(
        runs_std=("runs", "std"),
        strike_rate_std=("strike_rate", "std")
    )
)

display(batting_consistency.head(20))# ============================================
# BATTING CONSISTENCY
# ============================================

batting_consistency = (
    df
    .groupby("player", as_index=False)
    .agg(
        runs_std=("runs", "std"),
        strike_rate_std=("strike_rate", "std")
    )
)

display(batting_consistency.head(20))
```

```python
# ============================================
# BOWLING CONSISTENCY
# ============================================

bowling_consistency = (
    df
    .groupby("player", as_index=False)
    .agg(
        wickets_std=("wickets", "std"),
        economy_std=("economy", "std")
    )

)

display(bowling_consistency.head(20))
```

```python
print("Trend table:", first_last.shape)
print("Batting consistency:", batting_consistency.shape)
print("Bowling consistency:", bowling_consistency.shape)

print("\nTrend missing values:")
display(first_last.isna().sum())
```

```python
# ============================================
# PER-MATCH PERFORMANCE
# ============================================

df["runs_per_match"] = (
    df["runs"] /
    df["batting_matches"].replace(0, np.nan)
)

df["wickets_per_match"] = (
    df["wickets"] /
    df["bowling_matches"].replace(0, np.nan)
)

display(
    df[
        [
            "player",
            "season",
            "runs_per_match",
            "wickets_per_match"
        ]
    ].head(20)
)
```

```python
# ============================================
# PER-MATCH PERFORMANCE TREND
# ============================================

efficiency_trend = (
    df
    .sort_values(["player", "season"])
    .groupby("player")
    .agg(
        first_runs_per_match=("runs_per_match", "first"),
        latest_runs_per_match=("runs_per_match", "last"),
        first_wickets_per_match=("wickets_per_match", "first"),
        latest_wickets_per_match=("wickets_per_match", "last")
    )
    .reset_index()
)

efficiency_trend["runs_per_match_trend"] = (
    efficiency_trend["latest_runs_per_match"] -
    efficiency_trend["first_runs_per_match"]
)

efficiency_trend["wickets_per_match_trend"] = (
    efficiency_trend["latest_wickets_per_match"] -
    efficiency_trend["first_wickets_per_match"]
)

display(efficiency_trend.head(20))
```

```python
# ============================================
# COMBINE RECENT-FORM FEATURES
# ============================================

recent_form = latest_features.merge(
    recent_3_features,
    on="player",
    how="outer"
)

recent_form = recent_form.merge(
    first_last,
    on="player",
    how="outer"
)

recent_form = recent_form.merge(
    batting_consistency,
    on="player",
    how="outer"
)

recent_form = recent_form.merge(
    bowling_consistency,
    on="player",
    how="outer"
)

recent_form = recent_form.merge(
    efficiency_trend[
        [
            "player",
            "first_runs_per_match",
            "latest_runs_per_match",
            "first_wickets_per_match",
            "latest_wickets_per_match",
            "runs_per_match_trend",
            "wickets_per_match_trend"
        ]
    ],
    on="player",
    how="outer"
)

print("Shape:", recent_form.shape)
print("Players:", recent_form["player"].nunique())

display(recent_form.head(20))
```

```python
print(
    "Duplicate players:",
    recent_form["player"].duplicated().sum()
)
print("Missing player names:", recent_form["player"].isna().sum())
```

```python
# ============================================
# RECENCY WEIGHTS
# ============================================

season_weights = {
    2022: 0.10,
    2023: 0.15,
    2024: 0.20,
    2025: 0.25,
    2026: 0.30
}

print("Weights:", season_weights)
print("Weight total:", sum(season_weights.values()))
```

```python
# ============================================
# WEIGHTED RECENT PERFORMANCE
# ============================================

weighted_df = df.copy()

weighted_df["season_weight"] = (
    weighted_df["season"].map(season_weights)
)

weighted_df["weighted_runs"] = (
    weighted_df["runs"] *
    weighted_df["season_weight"]
)

weighted_df["weighted_wickets"] = (
    weighted_df["wickets"] *
    weighted_df["season_weight"]
)

weighted_performance = (
    weighted_df
    .groupby("player", as_index=False)
    .agg(
        weighted_runs=("weighted_runs", "sum"),
        weighted_wickets=("weighted_wickets", "sum")
    )
)

display(weighted_performance.head(20))
```

```python
weighted_df["weighted_balls"] = (
    weighted_df["batting_balls"] *
    weighted_df["season_weight"]
)

weighted_df["weighted_bowling_balls"] = (
    weighted_df["bowling_balls"] *
    weighted_df["season_weight"]
)

weighted_df["weighted_runs_conceded"] = (
    weighted_df["runs_conceded"] *
    weighted_df["season_weight"]
)

weighted_performance = (
    weighted_df
    .groupby("player", as_index=False)
    .agg(
        weighted_runs=("weighted_runs", "sum"),
        weighted_balls=("weighted_balls", "sum"),
        weighted_wickets=("weighted_wickets", "sum"),
        weighted_bowling_balls=("weighted_bowling_balls", "sum"),
        weighted_runs_conceded=("weighted_runs_conceded", "sum")
    )
)

weighted_performance["weighted_strike_rate"] = (
    weighted_performance["weighted_runs"] /
    weighted_performance["weighted_balls"].replace(0, np.nan)
) * 100

weighted_performance["weighted_economy"] = (
    weighted_performance["weighted_runs_conceded"] /
    (weighted_performance["weighted_bowling_balls"] / 6)
        .replace(0, np.nan)
)

display(weighted_performance.head(20))
```

```python
recent_form = recent_form.merge(
    weighted_performance,
    on="player",
    how="left"
)

print("Final recent-form shape:", recent_form.shape)
display(recent_form.head(10))
```

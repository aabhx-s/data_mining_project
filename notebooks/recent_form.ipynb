```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

ROOT = Path.cwd().resolve()
if not (ROOT / 'src').exists():
    raise RuntimeError('Open this notebook from the project root folder.')

from src.transformation.feature_builder import build_recent_form_features

AS_OF_SEASON = 2026
RECENT_SEASONS = 3
```

```python
recent_form = build_recent_form_features(
    ROOT,
    as_of_season=AS_OF_SEASON,
    recent_seasons=RECENT_SEASONS,
)
recent_form.head()
```

```python
display(recent_form.nlargest(15, 'recent_runs')[['player', 'recent_runs', 'recent_strike_rate', 'recent_batting_matches']])
display(recent_form.nlargest(15, 'recent_wickets')[['player', 'recent_wickets', 'recent_economy', 'recent_bowling_matches']])
```

```python
top_batters = recent_form.nlargest(10, 'recent_runs').sort_values('recent_runs')
top_batters.plot.barh(x='player', y='recent_runs', legend=False, figsize=(9, 5), color='#1f77b4', title=f'Top IPL run scorers through {AS_OF_SEASON}')
plt.xlabel('Runs across recent seasons')
plt.ylabel('Player')
plt.show()
```

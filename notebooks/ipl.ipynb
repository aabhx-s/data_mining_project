```python
from pathlib import Path
import sys

import pandas as pd

ROOT = Path.cwd().resolve()

while ROOT != ROOT.parent and not (ROOT / "src").exists():
    ROOT = ROOT.parent

if not (ROOT / "src").exists():
    raise RuntimeError("Open this notebook inside the project folder.")

if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from src.transformation.cricket_etl import DATASETS, run_default_pipeline
```

```python
processed_matches = run_default_pipeline(ROOT)
processed_matches.head()
```

```python
summary = []
for competition in DATASETS:
    folder = ROOT / 'data' / 'processed' / competition
    matches = pd.read_csv(folder / 'matches.csv')
    batting = pd.read_csv(folder / 'batting_match.csv')
    bowling = pd.read_csv(folder / 'bowling_match.csv')
    summary.append({
        'competition': competition.upper(),
        'matches': matches['match_id'].nunique(),
        'batting_rows': len(batting),
        'bowling_rows': len(bowling),
        'missing_player_ids': batting['player_id'].isna().sum() + bowling['player_id'].isna().sum(),
    })

pd.DataFrame(summary).sort_values('competition')
```

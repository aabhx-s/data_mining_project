```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

ROOT = Path.cwd().resolve()
if not (ROOT / 'src').exists():
    raise RuntimeError('Open this notebook from the project root folder.')

from src.transformation.player_master_etl import build_player_master
```

```python
player_master = build_player_master(ROOT)
player_master.head()
```

```python
quality = pd.DataFrame({
    'field': ['player_id', 'player', 'country', 'batting_style', 'bowling_style', 'playing_role'],
    'missing_values': [player_master[column].isna().sum() for column in ['player_id', 'player', 'country', 'batting_style', 'bowling_style', 'playing_role']]
})
quality
```

```python
roles = player_master['playing_role'].fillna('Unknown').value_counts().head(12).sort_values()
roles.plot.barh(figsize=(9, 5), color='#1f77b4', title='Most common player roles')
plt.xlabel('Players')
plt.ylabel('Role')
plt.show()

player_master[player_master['player'].str.contains('Kohli', case=False, na=False)][['player_id', 'player', 'country', 'playing_role']]
```

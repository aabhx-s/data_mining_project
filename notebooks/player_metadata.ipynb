```python
import pandas as pd

# Load player metadata
metadata_path = r"C:\stuff\DATA MINING PROJECT\data\raw\player_metadata.csv"
player_meta = pd.read_csv(metadata_path)

# people.csv
people_path = r"C:\stuff\DATA MINING PROJECT\data\raw\people.csv"
people = pd.read_csv(people_path)

# IPL batting data
batting_path = r"C:\stuff\DATA MINING PROJECT\data\processed\ipl\batting_match.csv"
batting = pd.read_csv(batting_path)

print("Player metadata:", player_meta.shape)
print("People:", people.shape)
print("IPL batting:", batting.shape)
```

```python
people_ids = set(
    people["identifier"]
    .dropna()
    .astype(str)
)

metadata_ids = set(
    player_meta["cricsheet_id"]
    .dropna()
    .astype(str)
)

overlap = people_ids & metadata_ids

print("People IDs:", len(people_ids))
print("Metadata IDs:", len(metadata_ids))
print("Matching IDs:", len(overlap))

print(
    "Match percentage:",
    round(len(overlap) / len(metadata_ids) * 100, 2),
    "%"
)
```

```python
ipl_players = (
    batting["player"]
    .dropna()
    .astype(str)
    .str.strip()
    .unique()
)

print("Unique IPL players:", len(ipl_players))

# Check exact matches against people.name
exact_name_matches = set(ipl_players) & set(people["name"].astype(str))

print("Exact matches with people.name:", len(exact_name_matches))
print(
    "Exact match percentage:",
    round(len(exact_name_matches) / len(ipl_players) * 100, 2),
    "%"
)
```

```python
display(
    people[
        people["name"].isin(ipl_players)
    ][
        ["identifier", "name", "unique_name", "key_cricinfo"]
    ].head(30)
)
```

```python
# Find IPL players that don't have an exact match in people.name

people_names = set(
    people["name"]
    .dropna()
    .astype(str)
    .str.strip()
)

unmatched_players = sorted(
    set(ipl_players) - people_names
)

print("Unmatched IPL players:", len(unmatched_players))
print(unmatched_players)

people_unique_names = set(
    people["unique_name"]
    .dropna()
    .astype(str)
    .str.strip()
)

unique_name_matches = set(unmatched_players) & people_unique_names

print("Matches in unique_name:", len(unique_name_matches))
print(unique_name_matches)
```

```python
# Create a clean lookup from player name → Cricsheet ID

people_lookup = people[
    ["identifier", "name", "unique_name"]
].copy()

# First try the normal name
name_to_id = dict(
    zip(
        people_lookup["name"].astype(str).str.strip(),
        people_lookup["identifier"].astype(str).str.strip()
    )
)

# Add unique_name as fallback
unique_name_to_id = dict(
    zip(
        people_lookup["unique_name"].astype(str).str.strip(),
        people_lookup["identifier"].astype(str).str.strip()
    )
)

# Map IPL player names to Cricsheet IDs
ipl_player_mapping = []

for player in sorted(ipl_players):
    if player in name_to_id:
        cricsheet_id = name_to_id[player]
        match_type = "name"
    elif player in unique_name_to_id:
        cricsheet_id = unique_name_to_id[player]
        match_type = "unique_name"
    else:
        cricsheet_id = None
        match_type = "unmatched"

    ipl_player_mapping.append({
        "player": player,
        "cricsheet_id": cricsheet_id,
        "match_type": match_type
    })

ipl_player_mapping = pd.DataFrame(ipl_player_mapping)

display(ipl_player_mapping.head(20))
```

```python
print(
    ipl_player_mapping["match_type"].value_counts()
)
```

```python
# Keep only the metadata fields we actually need
metadata_columns = [
    "cricsheet_id",
    "cricinfo_id",
    "unique_name",
    "full_name",
    "name",
    "country",
    "dob",
    "batting_style",
    "bowling_style",
    "birthplace",
    "playing_role"
]

metadata_clean = player_meta[metadata_columns].copy()

# Make sure the join keys have the same type
ipl_player_mapping["cricsheet_id"] = (
    ipl_player_mapping["cricsheet_id"]
    .astype(str)
    .str.strip()
)

metadata_clean["cricsheet_id"] = (
    metadata_clean["cricsheet_id"]
    .astype(str)
    .str.strip()
)

# Merge IPL players with metadata
player_master = ipl_player_mapping.merge(
    metadata_clean,
    on="cricsheet_id",
    how="left",
    validate="one_to_one"
)

print("Player master shape:", player_master.shape)

display(player_master.head(20))
```

```python
# Check duplicate Cricsheet IDs in player metadata

duplicates = (
    metadata_clean[
        metadata_clean["cricsheet_id"].duplicated(keep=False)
    ]
    .sort_values("cricsheet_id")
)

print("Duplicate cricsheet IDs:", duplicates["cricsheet_id"].nunique())
print("Rows involved:", len(duplicates))

display(duplicates.head(30))
print(
    "Duplicate IDs in IPL mapping:",
    ipl_player_mapping["cricsheet_id"].duplicated().sum()
)
```

```python
duplicate_ipl = ipl_player_mapping[
    ipl_player_mapping["cricsheet_id"].duplicated(keep=False)
].sort_values("cricsheet_id")

display(duplicate_ipl)
```

```python
duplicate_metadata_ids = set(
    duplicates["cricsheet_id"]
)

ipl_duplicate_metadata = player_master_ids = ipl_player_mapping[
    ipl_player_mapping["cricsheet_id"].isin(duplicate_metadata_ids)
].sort_values("cricsheet_id")

display(ipl_duplicate_metadata)
```

```python
duplicate_id = duplicate_ipl["cricsheet_id"].iloc[0]

print("Duplicate Cricsheet ID:", duplicate_id)

display(
    player_meta[
        player_meta["cricsheet_id"].astype(str).str.strip() == duplicate_id
    ]
)
```

```python
# Create one metadata record per Cricsheet ID
metadata_unique = (
    metadata_clean
    .drop_duplicates(subset="cricsheet_id", keep="first")
    .copy()
)

print("Metadata rows before:", len(metadata_clean))
print("Metadata rows after:", len(metadata_unique))

# Create one player master record per Cricsheet ID
player_master = (
    ipl_player_mapping
    .dropna(subset=["cricsheet_id"])
    .drop_duplicates(subset=["cricsheet_id"], keep="first")
    .merge(
        metadata_unique,
        on="cricsheet_id",
        how="left",
        validate="one_to_one"
    )
)

print("Player master shape:", player_master.shape)

display(player_master.head(20))
```

```python
import pandas as pd

# Paths relative to notebooks/season_features.ipynb
batting_path = "../data/processed/ipl/batting_match.csv"
bowling_path = "../data/processed/ipl/bowling_match.csv"

batting = pd.read_csv(batting_path)
bowling = pd.read_csv(bowling_path)

print("Batting data:", batting.shape)
print("Bowling data:", bowling.shape)
```

```python
print("IPL seasons:")
print(sorted(batting["season"].unique()))

print("\nNumber of seasons:", batting["season"].nunique())

print("\nBatting records by season:")
display(
    batting.groupby("season")
    .size()
    .reset_index(name="records")
)
```

```python
bowling_season = (
    bowling
    .groupby(["player", "season"], as_index=False)
    .agg(
        matches=("match_id", "nunique"),
        balls=("balls", "sum"),
        runs_conceded=("runs_conceded", "sum"),
        wickets=("wickets", "sum")
    )
)

bowling_season["overs"] = (
    bowling_season["balls"] / 6
)

bowling_season["economy"] = (
    bowling_season["runs_conceded"] /
    bowling_season["overs"]
)

bowling_season["bowling_strike_rate"] = (
    bowling_season["balls"] /
    bowling_season["wickets"]
)

display(bowling_season.head(20))
```

```python
print(
    "Zero-wicket bowling records:",
    (bowling_season["wickets"] == 0).sum()
)

print(
    "Missing bowling strike rates:",
    bowling_season["bowling_strike_rate"].isna().sum()
)
```

```python
# Keep only the most recent 5 IPL seasons

recent_seasons = ["2021", "2022", "2023", "2024", "2025"]

batting = batting[
    batting["season"].astype(str).isin(recent_seasons)
].copy()

bowling = bowling[
    bowling["season"].astype(str).isin(recent_seasons)
].copy()

print("Seasons being used:")
print(sorted(batting["season"].unique()))

print("\nBatting shape:", batting.shape)
print("Bowling shape:", bowling.shape)
```

```python
print("Batting records by season:")
display(
    batting.groupby("season")
    .size()
    .reset_index(name="records")
)

print("\nBowling records by season:")
display(
    bowling.groupby("season")
    .size()
    .reset_index(name="records")
)
```

```python
# ============================================
# BATTING FEATURES — 5 RECENT IPL SEASONS
# ============================================

batting_season = (
    batting
    .groupby(["player", "season"], as_index=False)
    .agg(
        batting_matches=("match_id", "nunique"),
        runs=("runs", "sum"),
        balls=("balls", "sum"),
        fours=("fours", "sum"),
        sixes=("sixes", "sum")
    )
)

# Strike rate
batting_season["strike_rate"] = (
    batting_season["runs"] /
    batting_season["balls"].replace(0, pd.NA)
) * 100

# Boundary runs
batting_season["boundary_runs"] = (
    batting_season["fours"] * 4 +
    batting_season["sixes"] * 6
)

# Percentage of scoring runs coming from boundaries
batting_season["boundary_percentage"] = (
    batting_season["boundary_runs"] /
    batting_season["runs"].replace(0, pd.NA)
) * 100

print("Shape:", batting_season.shape)

display(batting_season.head(20))
```

```python
print("Missing values:")
display(batting_season.isna().sum())

print("\nPlayers:", batting_season["player"].nunique())
print("Seasons:", batting_season["season"].nunique())
display(
    batting_season.sort_values(
        ["player", "season"]
    ).head(30)
)
```

```python
# ============================================
# BOWLING FEATURES FOR 5 RECENT IPL SEASONS
# ============================================

bowling_season = (
    bowling
    .groupby(["player", "season"], as_index=False)
    .agg(
        bowling_matches=("match_id", "nunique"),
        balls=("balls", "sum"),
        runs_conceded=("runs_conceded", "sum"),
        wickets=("wickets", "sum")
    )
)

# Overs
bowling_season["overs"] = (
    bowling_season["balls"] / 6
)

# Economy
bowling_season["economy"] = (
    bowling_season["runs_conceded"] /
    bowling_season["overs"].replace(0, pd.NA)
)

# Bowling strike rate
bowling_season["bowling_strike_rate"] = (
    bowling_season["balls"] /
    bowling_season["wickets"].replace(0, pd.NA)
)

print("Shape:", bowling_season.shape)

display(bowling_season.head(20))
```

```python
print("Missing values:")
display(bowling_season.isna().sum())

print("\nPlayers:", bowling_season["player"].nunique())
print("Seasons:", bowling_season["season"].nunique())
```
